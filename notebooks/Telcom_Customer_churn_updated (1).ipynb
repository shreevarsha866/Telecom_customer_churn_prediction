#Feature Engineering and Encoding

```python
x=df.drop('Churn',axis=1)
y=df['Churn'].map({'Yes':1,'No':0})
```

```python
x=x.drop('customerID',axis=1)  #drop unnecessary columns
```

```python
# Tenure groups
df['tenure_group'] = pd.cut(df['tenure'],
                            bins=[0,12,24,48,60,100],
                            labels=[1,2,3,4,5]).astype(int)

# Average spend
df['avg_spend'] = df['TotalCharges']/(df['tenure']+1)

# High value customer
df['high_value_customer'] = (df['MonthlyCharges'] >
                             df['MonthlyCharges'].median()).astype(int)

# Long term customer
df['long_term_customer'] = (df['tenure'] > 24).astype(int)

# Tenure charge ratio
df['tenure_charge_ratio'] = df['tenure']/(df['MonthlyCharges']+1)

# Contract encoding
contract_map = {'Month-to-month':1,'One year':2,'Two year':3}
df['contract_length'] = df['Contract'].map(contract_map)

# Premium internet
df['premium_internet'] = (df['InternetService']=='Fiber optic').astype(int)

# Paperless billing flag
df['paperless_flag'] = (df['PaperlessBilling']=='Yes').astype(int)

# Number of services
services = [
'PhoneService','OnlineSecurity','OnlineBackup',
'DeviceProtection','TechSupport','StreamingTV','StreamingMovies'
]

df['num_services'] = df[services].apply(lambda x:(x=='Yes').sum(),axis=1)
```

# PySpark Section — Distributed Data Processing

> **Why PySpark?** The preprocessing above uses pandas which works on a single machine. PySpark runs the **same logic on a distributed cluster**, making it production-ready for millions of records. This section replicates the feature engineering pipeline using Apache Spark (PySpark) and trains a model using Spark MLlib.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when, avg, count

# Initialize Spark Session
spark = SparkSession.builder \
    .appName('TelecomChurnAnalysis') \
    .config('spark.driver.memory', '2g') \
    .getOrCreate()

spark.sparkContext.setLogLevel('WARN')
print(f'Spark Version: {spark.version}')
print('Spark Session started successfully!')
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType

# Define schema explicitly, forcing TotalCharges to be StringType
schema = StructType([
    StructField("customerID", StringType(), True),
    StructField("gender", StringType(), True),
    StructField("SeniorCitizen", IntegerType(), True),
    StructField("Partner", StringType(), True),
    StructField("Dependents", StringType(), True),
    StructField("tenure", IntegerType(), True),
    StructField("PhoneService", StringType(), True),
    StructField("MultipleLines", StringType(), True),
    StructField("InternetService", StringType(), True),
    StructField("OnlineSecurity", StringType(), True),
    StructField("OnlineBackup", StringType(), True),
    StructField("DeviceProtection", StringType(), True),
    StructField("TechSupport", StringType(), True),
    StructField("StreamingTV", StringType(), True),
    StructField("StreamingMovies", StringType(), True),
    StructField("Contract", StringType(), True),
    StructField("PaperlessBilling", StringType(), True),
    StructField("PaymentMethod", StringType(), True),
    StructField("MonthlyCharges", DoubleType(), True),
    StructField("TotalCharges", StringType(), True), # Explicitly read as StringType
    StructField("Churn", StringType(), True)
])

# Load the CSV into a Spark DataFrame with the defined schema
spark_df = spark.read.csv(
    '/content/WA_Fn-UseC_-Telco-Customer-Churn.csv',
    header=True,
    schema=schema
)

print(f'Total records: {spark_df.count():,}')
print(f'Columns: {len(spark_df.columns)}')
spark_df.printSchema()
```

```python
from pyspark.sql.functions import col, when, udf
from pyspark.sql.types import DoubleType

# Define a UDF to safely cast string to double, returning None for malformed inputs
def safe_cast_to_double(value):
    try:
        return float(value)
    except (ValueError, TypeError):
        return None
safe_cast_udf = udf(safe_cast_to_double, DoubleType())

# Apply the UDF to TotalCharges and then drop rows where TotalCharges is NULL
spark_df = spark_df.withColumn('TotalCharges', safe_cast_udf(col('TotalCharges'))).dropna(subset=['TotalCharges'])

# Encode target: Yes=1, No=0
spark_df = spark_df.withColumn(
    'label',
    when(col('Churn') == 'Yes', 1).otherwise(0)
)

print(f'Clean dataset: {spark_df.count():,} rows')
spark_df.groupBy('label').count().show()
```

```python
# Quick EDA using Spark SQL — same insights as your pandas EDA
print('=== Churn by Contract Type ===')
spark_df.groupBy('Contract', 'label').count().orderBy('Contract').show()

print('=== Avg Tenure & Charges by Churn ===')
spark_df.groupBy('label').agg(
    avg('tenure').alias('avg_tenure'),
    avg('MonthlyCharges').alias('avg_monthly_charges'),
    count('*').alias('total_customers')
).show()
```

```python
from pyspark.sql.functions import when, col

# Feature Engineering — same features as your pandas section (Cell 34)
# Rewritten using Spark DataFrame transformations

spark_df = spark_df \
    .withColumn('tenure_group',
        when(col('tenure') <= 12, 1)
        .when(col('tenure') <= 24, 2)
        .when(col('tenure') <= 48, 3)
        .when(col('tenure') <= 60, 4)
        .otherwise(5)
    ) \
    .withColumn('avg_spend',
        col('TotalCharges') / (col('tenure') + 1)
    ) \
    .withColumn('long_term_customer',
        when(col('tenure') > 24, 1).otherwise(0)
    ) \
    .withColumn('tenure_charge_ratio',
        col('tenure') / (col('MonthlyCharges') + 1)
    ) \
    .withColumn('contract_length',
        when(col('Contract') == 'Month-to-month', 1)
        .when(col('Contract') == 'One year', 2)
        .otherwise(3)
    ) \
    .withColumn('premium_internet',
        when(col('InternetService') == 'Fiber optic', 1).otherwise(0)
    ) \
    .withColumn('paperless_flag',
        when(col('PaperlessBilling') == 'Yes', 1).otherwise(0)
    )

print('Feature engineering complete!')
spark_df.select('tenure_group', 'avg_spend', 'long_term_customer',
                'contract_length', 'premium_internet', 'label').show(5)
```

## Spark MLlib Pipeline

Spark MLlib is the distributed machine learning library — equivalent to scikit-learn but runs across a cluster. We use the same `Pipeline` concept: encode → assemble → scale → train.


```python
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler, StandardScaler
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import BinaryClassificationEvaluator
from pyspark.ml import Pipeline
from pyspark.ml.tuning import CrossValidator, ParamGridBuilder

# Step 1: Encode categorical columns (same as your get_dummies)
cat_cols = ['Contract', 'InternetService', 'PaymentMethod', 'gender']
indexers = [StringIndexer(inputCol=c, outputCol=f'{c}_idx', handleInvalid='keep')
            for c in cat_cols]
encoders = [OneHotEncoder(inputCol=f'{c}_idx', outputCol=f'{c}_enc')
            for c in cat_cols]

# Step 2: Assemble all features into one vector (required by Spark ML)
numeric_features = ['tenure', 'MonthlyCharges', 'TotalCharges', 'SeniorCitizen',
                    'tenure_group', 'avg_spend', 'long_term_customer',
                    'tenure_charge_ratio', 'contract_length',
                    'premium_internet', 'paperless_flag']
encoded_features = [f'{c}_enc' for c in cat_cols]

assembler = VectorAssembler(
    inputCols=numeric_features + encoded_features,
    outputCol='features_raw',
    handleInvalid='keep'
)

# Step 3: Scale (same as your StandardScaler)
scaler_spark = StandardScaler(
    inputCol='features_raw',
    outputCol='features',
    withStd=True, withMean=False
)

# Step 4: Random Forest (same model as your sklearn RF)
rf_spark = RandomForestClassifier(
    featuresCol='features',
    labelCol='label',
    numTrees=100,
    maxDepth=5,
    seed=42
)

# Step 5: Build pipeline
pipeline = Pipeline(stages=indexers + encoders + [assembler, scaler_spark, rf_spark])
print('Pipeline built with stages:', [type(s).__name__ for s in pipeline.getStages()])
```

```python
# Train-Test Split using Spark
train_spark, test_spark = spark_df.randomSplit([0.8, 0.2], seed=42)
print(f'Train: {train_spark.count():,} rows | Test: {test_spark.count():,} rows')

# Cross-Validation (same as your GridSearchCV — but distributed)
param_grid = ParamGridBuilder() \
    .addGrid(rf_spark.numTrees, [50, 100]) \
    .addGrid(rf_spark.maxDepth, [4, 6]) \
    .build()

evaluator = BinaryClassificationEvaluator(
    labelCol='label',
    metricName='areaUnderROC'
)

cv = CrossValidator(
    estimator=pipeline,
    estimatorParamMaps=param_grid,
    evaluator=evaluator,
    numFolds=3,
    seed=42
)

print('Training Spark ML model with cross-validation...')
cv_model = cv.fit(train_spark)
print('Training complete!')
```

```python
# Evaluate on test set
predictions = cv_model.transform(test_spark)

roc_auc = evaluator.evaluate(predictions)
print(f'Spark ML ROC-AUC: {roc_auc:.4f}')

# Compare with your sklearn results
print('\n=== Model Comparison ===')
print(f'Sklearn Logistic Regression AUC : 0.835  (from your pandas notebook)')
print(f'Sklearn Random Forest AUC       : 0.819  (from your pandas notebook)')
print(f'Spark MLlib Random Forest AUC   : {roc_auc:.3f}  (distributed pipeline)')

# Prediction breakdown
print('\nPrediction Distribution:')
predictions.groupBy('label', 'prediction').count().orderBy('label').show()
```

```python
# Feature Importance from Spark Random Forest
# Same concept as your Cell 77 (sklearn RF feature importance)

best_pipeline = cv_model.bestModel
rf_stage = best_pipeline.stages[-1]  # Last stage is the RF model

importances = rf_stage.featureImportances.toArray()
feature_names = numeric_features + cat_cols

import pandas as pd
import matplotlib.pyplot as plt

feat_imp_df = pd.DataFrame({
    'Feature': feature_names[:len(importances)],
    'Importance': importances[:len(feature_names)]
}).sort_values('Importance', ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.barh(feat_imp_df['Feature'][::-1], feat_imp_df['Importance'][::-1], color='#3498db')
plt.xlabel('Importance Score')
plt.title('Top 10 Features — Spark MLlib Random Forest', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print(feat_imp_df.to_string(index=False))
```

#Model Training and Evaluation

Train Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
log_model=LogisticRegression(max_iter=1000)
log_model.fit(x_train_scaled,y_train)
y_pred_log=log_model.predict(x_test_scaled)
```

#Model Optimization for Business Objectives

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression                   #hyperparameter tuning

param_grid={
    'C':[0.01,0.1,0.5,1,5,10],
    'penalty':['l2']
}
grid=GridSearchCV(
    LogisticRegression(max_iter=1000),
    param_grid,
    cv=5,
    scoring='roc_auc'
)

grid.fit(x_train_scaled,y_train)
print('Best Parameters:',grid.best_params_)
```

```python
best_log=grid.best_estimator_
y_pred_best=best_log.predict(x_test_scaled)
y_prob_best=best_log.predict_proba(x_test_scaled)[:,1]
```

```python
from sklearn.metrics import classification_report,roc_auc_score

print(classification_report(y_test,y_pred_best))
print('AUC:',roc_auc_score(y_test,y_prob_best))
```
